```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
import os
import glob

# --- 1. Load and preprocess dynamically from folder ---
def load_and_prepare(path, suffix):
    df = pd.read_csv(path, parse_dates=['valid_time'])
    df = df.sort_values('valid_time')
    df['tp'] *= 1000  # Convert m to mm
    df = df[['valid_time', 'u10', 'v10', 't2m', 'sp', 'tp']]
    df.columns = ['valid_time'] + [f'{col}_{suffix}' for col in df.columns if col != 'valid_time']
    return df

# New path for updated data
data_folder = 'brazil/'

# Find all CSV files
file_paths = glob.glob(os.path.join(data_folder, '*.csv'))

# Extract suffixes from file names (e.g. 'brazil/n2e.csv' -> 'n2e')
region_suffixes = [os.path.splitext(os.path.basename(f))[0] for f in file_paths]

# Load all regions
dfs_by_region = [load_and_prepare(path, suffix) for path, suffix in zip(file_paths, region_suffixes)]

# Merge all dataframes on 'valid_time'
df = dfs_by_region[0]
for regional_df in dfs_by_region[1:]:
    df = df.merge(regional_df, on='valid_time')

df = df.sort_values('valid_time').reset_index(drop=True)

# 2. Add cyclical time features
df['dayofyear'] = df['valid_time'].dt.dayofyear
df['dayofyear_sin'] = np.sin(2 * np.pi * df['dayofyear'] / 365)
df['dayofyear_cos'] = np.cos(2 * np.pi * df['dayofyear'] / 365)

# 3. Define feature groups
regions = region_suffixes

# Define grid positions (5x5 grid, 0-4 indexing, c at (2,2))
positions = {
    'c': (2, 2),
    'e': (2, 3),
    'w': (2, 1),
    'n': (1, 2),
    's': (3, 2),
    'ne': (1, 3),
    'nw': (1, 1),
    'se': (3, 3),
    'sw': (3, 1),
    'ne2': (0, 4),
    'nw2': (0, 0),
    'se2': (4, 4),
    'sw2': (4, 0),
    'n2': (0, 2),
    's2': (4, 2),
    'e2': (2, 4),
    'w2': (2, 0),
    # Add more if you have them, e.g., 'nn': (0, 2) if n2 is already there
}

# Ensure all regions in positions are in your data
regions = [r for r in regions if r in positions]

# 4. Collect all meteorological input features from all regions
all_meteorological_features = []
for region in regions:
    all_meteorological_features += [f'{var}_{region}' for var in ['u10', 'v10', 't2m', 'sp', 'tp']]

# 5. Add spatial gradient features around 'c' to capture frontal movements
# Neighbors of 'c': direct (e, w, n, s) and diagonal (ne, nw, se, sw)
neighbors = {
    'e': 'e',
    'w': 'w',
    'n': 'n',
    's': 's',
    'ne': 'ne',
    'nw': 'nw',
    'se': 'se',
    'sw': 'sw'
}

diff_features = []
for var in ['u10', 'v10', 't2m', 'sp', 'tp']:
    for dir_key, neigh in neighbors.items():
        if neigh in regions:
            diff_col = f'{var}_c_d{dir_key}'  # e.g., t2m_c_de = t2m_c - t2m_e
            df[diff_col] = df[f'{var}_c'] - df[f'{var}_{neigh}']
            diff_features.append(diff_col)

# All input features: meteorological + time + spatial differences
all_input_features = all_meteorological_features + ['dayofyear_sin', 'dayofyear_cos'] + diff_features

# Add rolling features
df['rolling_mean_7d'] = df['tp_c'].rolling(window=7*24, min_periods=1).mean()  # 7 days average
df['rolling_max_24h'] = df['tp_c'].rolling(window=24, min_periods=1).max()  # Max in last 24 hours
df['spike_flag'] = (df['rolling_max_24h'] > 5).astype(int)  # 1 if spike likely

# Update all_input_features to include these
all_input_features += ['rolling_mean_7d', 'rolling_max_24h', 'spike_flag']

target = 'tp_c'
```

```python
from sklearn.model_selection import train_test_split
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

sequence_length = 24

# Aggregate tp_c into daily totals (assuming hourly data, group by date)
df['date'] = df['valid_time'].dt.date  # Extract date
daily_rain = df.groupby('date')['tp_c'].sum().reset_index()  # Sum tp_c per day
daily_rain.columns = ['date', 'daily_total_mm']

# Define classes (bins)
bins = [0, 1, 5, 10, 20, 50, float('inf')]  # Adjust as needed
labels = ['<1 mm', '1-5 mm', '5-10 mm', '10-20 mm', '20-50 mm', '>50 mm']
daily_rain['class'] = pd.cut(daily_rain['daily_total_mm'], bins=bins, labels=labels, right=False)


# Assuming daily_rain has 'date' and 'class'
# First, chronological split on dates
dates = daily_rain['date'].unique()
train_dates, temp_dates = train_test_split(dates, test_size=0.3, shuffle=False)  # 70% train
val_dates, test_dates = train_test_split(temp_dates, test_size=0.333, shuffle=False)  # 20% val, 10% test

# Filter df by dates
train_df = df[df['date'].isin(train_dates)]
val_df = df[df['date'].isin(val_dates)]
test_df = df[df['date'].isin(test_dates)]

# Oversample high-rain days in train (e.g., duplicate >20 mm days)
high_rain_dates = daily_rain[(daily_rain['date'].isin(train_dates)) & (daily_rain['class'].isin(['20-50 mm', '>50 mm']))]['date']
oversampled_train = pd.concat([train_df] + [train_df[train_df['date'].isin(high_rain_dates)]] * 2)  # Duplicate 2x

# Function to create sequences from a df
def create_sequences(df, all_input_features, target, sequence_length):
    X = []
    y = []
    for i in range(len(df) - sequence_length):
        X.append(df[all_input_features].iloc[i:i+sequence_length].values)
        y.append(df[target].iloc[i+sequence_length])
    return np.array(X), np.array(y)

# Create sequences for each split
X_train, y_train = create_sequences(oversampled_train, all_input_features, target, sequence_length)
X_val, y_val = create_sequences(val_df, all_input_features, target, sequence_length)
X_test, y_test = create_sequences(test_df, all_input_features, target, sequence_length)

# For 7-day daily totals: Modify target to sum of next 24 hours
days_ahead = 7
models = {}
scalers_X = {}
scalers_y = {}

for day in range(1, days_ahead + 1):
    print(f"Training model for Day {day}")
    
    # Target: Sum of tp_c over the next 24 hours (shifted by (day-1)*24)
    # Use the respective df for each split to compute y_daily
    y_train_daily = []
    for i in range(len(oversampled_train) - sequence_length - 24 * day):
        start_idx = i + sequence_length + (day - 1) * 24
        end_idx = start_idx + 24
        y_train_daily.append(oversampled_train[target].iloc[start_idx:end_idx].sum())
    y_train_daily = np.array(y_train_daily)
    
    y_val_daily = []
    for i in range(len(val_df) - sequence_length - 24 * day):
        start_idx = i + sequence_length + (day - 1) * 24
        end_idx = start_idx + 24
        y_val_daily.append(val_df[target].iloc[start_idx:end_idx].sum())
    y_val_daily = np.array(y_val_daily)
    
    y_test_daily = []
    for i in range(len(test_df) - sequence_length - 24 * day):
        start_idx = i + sequence_length + (day - 1) * 24
        end_idx = start_idx + 24
        y_test_daily.append(test_df[target].iloc[start_idx:end_idx].sum())
    y_test_daily = np.array(y_test_daily)
    
    # Adjust X to match lengths
    X_train_daily = X_train[:len(y_train_daily)]
    X_val_daily = X_val[:len(y_val_daily)]
    X_test_daily = X_test[:len(y_test_daily)]
    
    # Scale
    scaler_X = StandardScaler()
    scaler_y = StandardScaler()
    X_train_scaled = scaler_X.fit_transform(X_train_daily.reshape(-1, len(all_input_features))).reshape(X_train_daily.shape)
    y_train_scaled = scaler_y.fit_transform(y_train_daily.reshape(-1, 1))
    X_val_scaled = scaler_X.transform(X_val_daily.reshape(-1, len(all_input_features))).reshape(X_val_daily.shape)
    y_val_scaled = scaler_y.transform(y_val_daily.reshape(-1, 1))
    X_test_scaled = scaler_X.transform(X_test_daily.reshape(-1, len(all_input_features))).reshape(X_test_daily.shape)
    y_test_scaled = scaler_y.transform(y_test_daily.reshape(-1, 1))
    
    # Build model
    model = Sequential()
    model.add(LSTM(64, input_shape=(sequence_length, len(all_input_features)), return_sequences=True))
    model.add(Dropout(0.2))
    model.add(LSTM(32))
    model.add(Dropout(0.2))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mse', metrics=['mae'])
    
    # Train
    model.fit(X_train_scaled, y_train_scaled, epochs=50, batch_size=32, validation_data=(X_val_scaled, y_val_scaled), verbose=0)
    
    # Evaluate
    test_loss, test_mae = model.evaluate(X_test_scaled, y_test_scaled, verbose=0)
    print(f"  Day {day} Test MSE: {test_loss:.4f}, MAE: {test_mae:.4f}")
    
    # Store
    models[day] = model
    scalers_X[day] = scaler_X
    scalers_y[day] = scaler_y

# 7. Make a 7-day forecast (using the last sequence from test_df or full df— adjust as needed)
last_df = test_df  # Or df if you want the very last
last_X, _ = create_sequences(last_df, all_input_features, target, sequence_length)
last_sequence = last_X[-1].reshape(1, sequence_length, -1)
forecast = {}
for day in range(1, days_ahead + 1):
    scaler_X = scalers_X[day]
    scaler_y = scalers_y[day]
    model = models[day]
    
    last_scaled = scaler_X.transform(last_sequence.reshape(-1, len(all_input_features))).reshape(last_sequence.shape)
    pred_scaled = model.predict(last_scaled)
    pred_actual = scaler_y.inverse_transform(pred_scaled)
    forecast[f"Day {day}"] = pred_actual[0][0]

print("\n7-Day Rainfall Forecast (mm):")
for day, amount in forecast.items():
    print(f"{day}: {amount:.2f} mm")
```
