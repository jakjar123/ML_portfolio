```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

# Your existing load_and_prepare function (unchanged)
def load_and_prepare(path, suffix):
    df = pd.read_csv(path, parse_dates=['valid_time'])
    df = df.sort_values('valid_time')
    df['tp'] *= 1000  # m to mm
    df['tp'] = np.log1p(df['tp'])  # log transform precipitation
    df = df[['valid_time', 'u10', 'v10', 't2m', 'sp', 'tp']]
    df.columns = ['valid_time'] + [f'{col}_{suffix}' for col in df.columns if col != 'valid_time']
    return df

# Load and merge data (your existing code, unchanged)
df_center = load_and_prepare('data/brazil.csv', 'center')
df_north = load_and_prepare('data/brazil_north.csv', 'north')
df_south = load_and_prepare('data/brazil_south.csv', 'south')
df_east = load_and_prepare('data/brazil_east.csv', 'east')
df_west = load_and_prepare('data/brazil_west.csv', 'west')

df = df_center.merge(df_north, on='valid_time')
df = df.merge(df_south, on='valid_time')
df = df.merge(df_east, on='valid_time')
df = df.merge(df_west, on='valid_time')
df = df.sort_values('valid_time').reset_index(drop=True)

# Add cyclical time features (your existing code, unchanged)
df['dayofyear'] = df['valid_time'].dt.dayofyear
df['dayofyear_sin'] = np.sin(2 * np.pi * df['dayofyear'] / 365)
df['dayofyear_cos'] = np.cos(2 * np.pi * df['dayofyear'] / 365)

# Prepare features and target
features = df.drop(columns=['valid_time', 'tp_center'])  # Exclude target and time
target = df['tp_center']  # Log-transformed daily precipitation

# Scale features
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

# Create sequences
seq_length = 24  # 24 days input
pred_length = 7  # Predict 7 days ahead
X, y = [], []
for i in range(len(features_scaled) - seq_length - pred_length + 1):
    X.append(features_scaled[i:i + seq_length])
    y.append(target.iloc[i + seq_length:i + seq_length + pred_length].values)

X = np.array(X)
y = np.array(y)

# Split into train/val/test (temporal split: 70/15/15)
train_size = int(0.7 * len(X))
val_size = int(0.15 * len(X))
test_size = len(X) - train_size - val_size

X_train, X_val, X_test = X[:train_size], X[train_size:train_size + val_size], X[train_size + val_size:]
y_train, y_val, y_test = y[:train_size], y[train_size:train_size + val_size], y[train_size + val_size:]

# Build and train LSTM model
model = Sequential()
model.add(LSTM(50, input_shape=(seq_length, features.shape[1])))
model.add(Dense(pred_length))  # Output 7 predictions
model.compile(optimizer='adam', loss='mse')
model.fit(X_train, y_train, epochs=50, batch_size=32, validation_data=(X_val, y_val), verbose=1)

# Predict on test set
y_pred = model.predict(X_test)

# Inverse transform to original mm scale for evaluation (since target was log-transformed)
y_test_orig = np.expm1(y_test)  # expm1 because log1p was used
y_pred_orig = np.expm1(y_pred)

# Evaluate for each of the 7 days
for j in range(pred_length):
    r2 = r2_score(y_test_orig[:, j], y_pred_orig[:, j])
    mae = mean_absolute_error(y_test_orig[:, j], y_pred_orig[:, j])
    mse = mean_squared_error(y_test_orig[:, j], y_pred_orig[:, j])
    print(f"Day {j+1}: R2={r2:.4f}, MAE={mae:.4f}, MSE={mse:.4f}")
```
